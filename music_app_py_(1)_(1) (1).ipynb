```python
import os
import glob
import random
import numpy as np
from tqdm import tqdm
import pretty_midi
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Embedding
from midiutil import MIDIFile
import subprocess
from IPython.display import Audio, display

# Optional: sentence-transformers if you want text embeddings (not required for this LSTM demo)
# from sentence_transformers import SentenceTransformer

# ---------- Config ----------
WORKDIR = "/content/text2music"
os.makedirs(WORKDIR, exist_ok=True)

# Data/model params (tweak as desired)
NUM_EXAMPLES = 300     # how many .midi files to use from MAESTRO (or other folder)
MAX_NOTES_PER_FILE = 256  # how many notes to extract per file (truncate)
MIN_PITCH = 48         # C3
MAX_PITCH = 84         # C6
SEQ_LEN = 50           # context length for training
BATCH_SIZE = 16
EPOCHS = 8             # small for demo; increase for better results
EMBED_DIM = 128
LSTM_UNITS = 256

MAESTRO_ZIP_ON_DRIVE = "/content/drive/MyDrive/maestro-v3.0.0-midi.zip"  # update if needed
EXTRACT_DIR = "/content/maestro"  # where MAESTRO will be unzipped or scanned

SOUNDFONT = "/content/GeneralUser.sf2"
print("Config ready. WORKDIR:", WORKDIR)
```

```python
def midi_to_monophonic_notes(midi_path, min_pitch=MIN_PITCH, max_pitch=MAX_PITCH, max_notes=MAX_NOTES_PER_FILE):
    """
    Parse MIDI file with pretty_midi and return a list of pitch ints (monophonic-ish).
    We sort notes by start time and keep notes within pitch range. We collapse immediate repeats.
    """
    try:
        pm = pretty_midi.PrettyMIDI(midi_path)
    except Exception as e:
        # parsing error
        return []
    notes = []
    for inst in pm.instruments:
        # prefer piano-like instruments (program 0) if available — but include others
        for n in inst.notes:
            notes.append((n.start, n.pitch))
    if not notes:
        return []
    notes.sort(key=lambda x: x[0])
    pitches = []
    last_pitch = None
    for s, p in notes:
        if p < min_pitch or p > max_pitch:
            continue
        if last_pitch is None or p != last_pitch:
            pitches.append(p)
            last_pitch = p
        if len(pitches) >= max_notes:
            break
    return pitches

def save_pitches_to_midi(pitches, out_path="generated.mid", seconds_per_note=0.4, program=0):
    """
    Write a simple monophonic MIDI from integer pitches using pretty_midi.
    """
    pm = pretty_midi.PrettyMIDI()
    inst = pretty_midi.Instrument(program=program)
    t = 0.0
    for p in pitches:
        note = pretty_midi.Note(velocity=100, pitch=int(p), start=t, end=t + seconds_per_note)
        inst.notes.append(note)
        t += seconds_per_note
    pm.instruments.append(inst)
    pm.write(out_path)
    print("Saved MIDI:", out_path)
    return out_path

def midi_to_wav(midi_path, wav_path="generated.wav", soundfont=SOUNDFONT, sample_rate=44100):
    """
    Use fluidsynth to render the MIDI to a WAV using provided soundfont.
    Requires fluidsynth + soundfont installed.
    """
    if not os.path.exists(soundfont):
        print("Soundfont not found at", soundfont)
        return None
    cmd = ["fluidsynth", "-ni", soundfont, midi_path, "-F", wav_path, "-r", str(sample_rate)]
    try:
        subprocess.run(cmd, check=True, stdout=subprocess.PIPE, stderr=subprocess.PIPE)
        print("Rendered WAV:", wav_path)
        return wav_path
    except subprocess.CalledProcessError as e:
        print("fluidsynth failed:", e)
        return None
```

```python
# Choose folder to scan for MIDI files. If you extracted MAESTRO, EXTRACT_DIR is good; otherwise
# put your MIDI files in /content/midi_files and change DATA_DIR accordingly.

DATA_DIR = EXTRACT_DIR  # change if your MIDIs are elsewhere
midi_paths = glob.glob(os.path.join(DATA_DIR, "**/*.midi"), recursive=True) + \
             glob.glob(os.path.join(DATA_DIR, "**/*.mid"), recursive=True)
midi_paths = sorted(list(set(midi_paths)))
print("Found MIDI files:", len(midi_paths))

# If no MIDI files found, create toy data (simple repeating scales) so the rest runs.
if len(midi_paths) == 0:
    print("No MIDI files found. Creating toy data inside /content/midi_sample.")
    os.makedirs("/content/midi_sample", exist_ok=True)
    # create simple sequences as fallback
    for i in range(10):
        pitches = [60 + ((j + i) % 7) for j in range(100)]
        pth = f"/content/midi_sample/sample_{i}.mid"
        save_pitches_to_midi(pitches, pth, seconds_per_note=0.25)
    DATA_DIR = "/content/midi_sample"
    midi_paths = glob.glob(os.path.join(DATA_DIR, "*.mid"))

# Limit to NUM_EXAMPLES for speed
midi_paths = midi_paths[:NUM_EXAMPLES]
print("Using", len(midi_paths), "files for extraction.")

# Extract monophonic pitch sequences
all_sequences = []
for p in tqdm(midi_paths, desc="Parsing MIDIs"):
    seq = midi_to_monophonic_notes(p, max_notes=MAX_NOTES_PER_FILE)
    if len(seq) >= 4:  # keep sequences with at least a few notes
        all_sequences.append(seq)

print("Extracted sequences:", len(all_sequences))
# show some stats
lens = [len(s) for s in all_sequences]
print("Average length:", (sum(lens)/len(lens)) if lens else 0, "min/max:", (min(lens), max(lens)) if lens else (0,0))
```

```python
# Flatten notes to build vocab within [MIN_PITCH, MAX_PITCH]
flattened = [p for seq in all_sequences for p in seq]
if len(flattened) == 0:
    raise RuntimeError("No notes extracted. Check DATA_DIR or MIDI parsing.")

unique_pitches = sorted(list(set(flattened)))
print("Unique pitches in dataset:", unique_pitches)

pitch_to_int = {p: i for i, p in enumerate(unique_pitches)}
int_to_pitch = {i: p for p, i in pitch_to_int.items()}
vocab_size = len(unique_pitches)
print("Vocab size:", vocab_size)

# Make sequences for next-token prediction (sliding window across concatenated sequences)
encoded_flat = []
for seq in all_sequences:
    encoded_flat.extend([pitch_to_int[p] for p in seq])

# Create (input_seq, target) pairs
inputs = []
targets = []
for i in range(0, len(encoded_flat) - SEQ_LEN):
    inputs.append(encoded_flat[i:i+SEQ_LEN])
    targets.append(encoded_flat[i+SEQ_LEN])

inputs = np.array(inputs, dtype=np.int32)
targets = np.array(targets, dtype=np.int32)

print("Prepared training data:", inputs.shape, "->", targets.shape)

# If dataset too large, subsample
MAX_TRAIN_SAMPLES = 20000
if inputs.shape[0] > MAX_TRAIN_SAMPLES:
    idx = np.random.choice(range(inputs.shape[0]), size=MAX_TRAIN_SAMPLES, replace=False)
    inputs = inputs[idx]
    targets = targets[idx]
    print("Subsampled to", inputs.shape[0], "training samples")
```

```python
model = Sequential([
    Embedding(input_dim=vocab_size, output_dim=EMBED_DIM, input_length=SEQ_LEN),
    LSTM(LSTM_UNITS, return_sequences=True),
    LSTM(LSTM_UNITS),
    Dense(LSTM_UNITS, activation="relu"),
    Dense(vocab_size, activation="softmax")
])

model.build(input_shape=(None, SEQ_LEN))
model.summary()

model.compile(
    loss="sparse_categorical_crossentropy",
    optimizer=tf.keras.optimizers.Adam(1e-3),
    metrics=["accuracy"]
)

history = model.fit(
    inputs, targets,
    epochs=EPOCHS,
    batch_size=BATCH_SIZE,
    validation_split=0.1,
    verbose=1
)
```

```python
def sample(preds, temperature=1.0, top_k=None):
    """
    preds: 1D array of probabilities (softmax output)
    """
    preds = np.asarray(preds).astype("float64")
    if top_k is not None and top_k > 0:
        # zero out everything except top_k
        topk_idx = np.argpartition(preds, -top_k)[-top_k:]
        mask = np.ones_like(preds, dtype=bool)
        mask[topk_idx] = False
        preds[mask] = 0.0
    preds = preds / (preds.sum() + 1e-12)
    # avoid log(0)
    log_preds = np.log(preds + 1e-20) / max(temperature, 1e-9)
    exp_preds = np.exp(log_preds)
    probs = exp_preds / (np.sum(exp_preds) + 1e-12)
    return np.random.choice(len(probs), p=probs)

def generate_sequence(model, seed_indices, length=200, temperature=1.0, top_k=20, no_repeat_n=3):
    generated = list(seed_indices)
    for _ in range(length):
        x = np.array(generated[-SEQ_LEN:]).reshape(1, SEQ_LEN)
        preds = model.predict(x, verbose=0)[0]
        # penalize immediate repeats
        for back in range(1, no_repeat_n+1):
            if len(generated) - back >= 0:
                preds[generated[-back]] *= 0.6
        next_idx = sample(preds, temperature=temperature, top_k=top_k)
        generated.append(int(next_idx))
    return generated
```

```python
# ============================================================
# 🎶 Improved Final Music Generation (Fixed Version)
# Author: Dakshani
# ============================================================

import os, random, time
import numpy as np
from IPython.display import Audio, display

# --- Safety: rebuild vocab if missing (ensures smooth run) ---
if 'unique_pitches' not in globals() or not unique_pitches:
    print("⚠️ unique_pitches missing — rebuilding default pitch range C3–C6")
    unique_pitches = list(range(48, 73))  # C3..C6 inclusive
    pitch_to_int = {p: i for i, p in enumerate(unique_pitches)}
    int_to_pitch = {i: p for i, p in enumerate(unique_pitches)}

vocab_size = len(unique_pitches)
MIN_PITCH, MAX_PITCH = min(unique_pitches), max(unique_pitches)

# --- Example Tamil / Ilaiyaraaja-style base motif (edit freely) ---
base_manual_motif = [60, 62, 64, 65, 67, 69, 71, 72]

# --- Safe pitch→index mapper (no KeyError ever) ---
def snap_to_vocab(pitch):
    if pitch in pitch_to_int:
        return pitch_to_int[pitch]
    # nearest pitch in vocab; fallback to first index if still missing
    nearest = min(unique_pitches, key=lambda x: abs(x - pitch))
    return pitch_to_int.get(nearest, list(pitch_to_int.values())[0])

# --- Mixed seed builder to add variation ---
def build_mixed_seed(inputs, seq_len, num_windows=3):
    """Concatenate small slices from random windows to form a varied seed"""
    seed_parts = []
    if inputs.shape[0] > 0:
        for _ in range(num_windows):
            idx = np.random.randint(0, inputs.shape[0])
            part_len = max(4, seq_len // num_windows)
            start = np.random.randint(0, max(1, inputs.shape[1] - part_len + 1))
            part = inputs[idx, start:start + part_len].tolist()
            seed_parts.extend(part)
        # clamp/trim/pad to exactly seq_len
        if len(seed_parts) >= seq_len:
            seed = seed_parts[-seq_len:]
        else:
            pad = [seed_parts[0]] * (seq_len - len(seed_parts))
            seed = pad + seed_parts
    else:
        # fallback: middle C or zeros
        middle_c_idx = pitch_to_int.get(60, 0)
        seed = [middle_c_idx] * seq_len
    return seed

# --- Build generation seed ---
seed = build_mixed_seed(inputs, SEQ_LEN, num_windows=3)

# --- Random transposition for variety ---
transpose_semitones = random.choice([-12, -7, -5, 0, 5, 7, 12])
transposed_motif = [
    max(MIN_PITCH, min(MAX_PITCH, p + transpose_semitones))
    for p in base_manual_motif
]
motif_indices = [snap_to_vocab(p) for p in transposed_motif]

# --- Merge motif into seed ---
if len(motif_indices) < SEQ_LEN:
    seed = seed[: (SEQ_LEN - len(motif_indices))] + motif_indices
else:
    seed = motif_indices[-SEQ_LEN:]

# --- Clamp to valid vocab range ---
seed = [int(s) if 0 <= int(s) < vocab_size else 0 for s in seed]

# --- Sampling hyperparameters ---
temperature = float(np.round(random.uniform(0.7, 1.5), 2))
top_k = random.choice([5, 10, 20, max(1, vocab_size - 1)])
no_repeat_n = random.choice([2, 3, 4])
length = random.choice([150, 200, 250])

print("Seed (first 30 indices):", seed[:30])
print("Transposed motif (MIDI):", transposed_motif)
print(
    f"Sampling params → temperature: {temperature}, top_k: {top_k}, "
    f"no_repeat_n: {no_repeat_n}, length: {length}"
)

# --- Generate sequence ---
gen_indices = generate_sequence(
    model,
    seed,
    length=length,
    temperature=temperature,
    top_k=top_k,
    no_repeat_n=no_repeat_n
)

# --- Convert indices → MIDI pitches safely ---
def idx_to_pitch_safe(idx):
    if idx in int_to_pitch:
        return int_to_pitch[idx]
    idx_clamped = max(0, min(idx, max(int_to_pitch.keys())))
    return int_to_pitch[idx_clamped]

generated_pitches = [idx_to_pitch_safe(int(i)) for i in gen_indices]
print("Generated (first 40 pitches):", generated_pitches[:40])

# --- Save & render ---
seconds_per_note = round(random.uniform(0.28, 0.42), 3)
timestamp = int(time.time())
midi_path = os.path.join(WORKDIR, f"generated_varied_{timestamp}.mid")
wav_path = os.path.join(WORKDIR, f"generated_varied_{timestamp}.wav")

save_pitches_to_midi(generated_pitches, midi_path, seconds_per_note=seconds_per_note)
print(f"🎼 Saved MIDI: {midi_path}, seconds_per_note: {seconds_per_note}")

# --- Optional WAV rendering if fluidsynth is available ---
if os.path.exists(SOUNDFONT):
    out = midi_to_wav(midi_path, wav_path, soundfont=SOUNDFONT)
    if out and os.path.exists(wav_path):
        display(Audio(wav_path, autoplay=False))
    else:
        print("⚠️ WAV not produced; check fluidsynth output.")
else:
    print("⚠️ SoundFont missing; skipping WAV rendering.")
```

```python
# ==========================================================
# ⚡ Text-to-Music LSTM Generator: Optimized for Speed
# ==========================================================
from google.colab import files
import os, shutil, zipfile, sys
import pickle, numpy as np, tensorflow as tf, music21 as m21
from tensorflow import keras
from tensorflow.keras import layers
from concurrent.futures import ProcessPoolExecutor, as_completed # Import as_completed for timed stop

# --- Configuration (Optimized) ---
TARGET_DIR = "/content/midi_dataset/maestro-v3.0.0/"
MIDI_PATH = TARGET_DIR
MODEL_PATH = "/content/text2music/trained_model.h5"
sequence_length = 30 # Reduced from 50
max_workers = 4
MAX_FILES = 100 # <--- MAJOR SPEED BOOST: Only process this many files

# --- Clean up and Setup ---
print("🛠️ Setting up environment and cleaning previous runs...")
shutil.rmtree("/content/midi_dataset", ignore_errors=True)
shutil.rmtree("/content/text2music", ignore_errors=True)
os.makedirs("/content/midi_dataset", exist_ok=True)
os.makedirs("/content/text2music", exist_ok=True)

# -----------------------------------------------------------
# --- PHASE 1: UPLOAD AND EXTRACTION --- (No change needed here)
# -----------------------------------------------------------

print("\n-------------------------------------------")
print("📂 PHASE 1: UPLOAD AND EXTRACTION")
print("-------------------------------------------")

# 1. Prompt for Upload
print("🚨 **ACTION REQUIRED:** Please upload your maestro-v3.0.0-midi.zip file now...")
uploaded = files.upload()

if not uploaded:
    print("❌ ERROR: No file uploaded. Exiting.")
    sys.exit(0)

uploaded_file_name = list(uploaded.keys())[0]
ZIP_PATH = f"/content/midi_dataset/{uploaded_file_name}"
shutil.move(uploaded_file_name, ZIP_PATH)
print(f"✅ File saved to: {ZIP_PATH}")

# 2. Extraction
shutil.rmtree(TARGET_DIR, ignore_errors=True)
os.makedirs(TARGET_DIR, exist_ok=True)

try:
    print(f"📂 Extracting {os.path.basename(ZIP_PATH)} to /content/temp_maestro/...")

    with zipfile.ZipFile(ZIP_PATH, "r") as zip_ref:
        zip_ref.extractall("/content/temp_maestro/")

    source_dir = "/content/temp_maestro/maestro-v3.0.0/"
    if not os.path.exists(source_dir):
        source_dir = "/content/temp_maestro/"

    for item in os.listdir(source_dir):
        shutil.move(os.path.join(source_dir, item), os.path.join(TARGET_DIR, item))

    shutil.rmtree("/content/temp_maestro/", ignore_errors=True)
    os.remove(ZIP_PATH)

    midi_count = sum([len(files) for r, d, files in os.walk(TARGET_DIR) for f in files if f.endswith(('.mid', '.midi'))])

    if midi_count == 0:
        print("❌ CRITICAL: No MIDI files found after extraction. Check ZIP file contents.")
        sys.exit(0)
    print(f"✅ Extraction complete. Found {midi_count} MIDI files for training.")

except Exception as e:
    print(f"❌ An error occurred during extraction: {e}")
    sys.exit(0)


# -----------------------------------------------------------
# --- PHASE 2: MODEL TRAINING (Optimized Functions) ---
# -----------------------------------------------------------
print("\n-------------------------------------------")
print("🧠 PHASE 2: MODEL TRAINING (Optimized)")
print("-------------------------------------------")

# --- Helper Functions (Optimized for File Count) ---
def process_single_file(full_path):
    # (Same music parsing logic as before)
    notes = []
    if full_path.endswith((".midi", ".mid")):
        try:
            midi = m21.converter.parse(full_path)
            notes_to_parse = m21.instrument.partitionByInstrument(midi).parts[0].recurse() if m21.instrument.partitionByInstrument(midi) and len(m21.instrument.partitionByInstrument(midi).parts) > 0 else midi.flat.notes
            for element in notes_to_parse:
                if isinstance(element, m21.note.Note):
                    notes.append(str(element.pitch))
                elif isinstance(element, m21.chord.Chord):
                    notes.append('.'.join(str(n) for n in element.normalOrder))
        except Exception:
            pass
    return notes

def extract_notes_parallel(midi_dir, max_workers, max_files):
    # Get all MIDI files and limit to MAX_FILES
    all_midi_files = [os.path.join(root, file) for root, _, files in os.walk(midi_dir) for file in files if file.endswith((".midi", ".mid"))]
    all_midi_files = all_midi_files[:max_files] # <--- LIMIT APPLIED HERE

    if not all_midi_files: return []
    print(f"Found {len(all_midi_files)} files. Starting parallel extraction (Processing {len(all_midi_files)} files only).")

    all_notes = []
    with ProcessPoolExecutor(max_workers=max_workers) as executor:
        # Use as_completed for safety, though map is fine for this task
        results = list(executor.map(process_single_file, all_midi_files))
    for notes_list in results:
        all_notes.extend(notes_list)
    return all_notes

# --- Execution ---
notes = extract_notes_parallel(MIDI_PATH, max_workers, MAX_FILES) # Pass the limit

if not notes:
    print("❌ ERROR: No notes extracted. Cannot train.")
    sys.exit(0)

pitches = sorted(set(notes))
with open("vocab.pkl", "wb") as f: pickle.dump(pitches, f)
print(f"✅ Vocab created. Unique pitches: {len(pitches)}")

# Prepare sequences
note_to_int = {n: i for i, n in enumerate(pitches)}
inputs, outputs = [], []
if len(notes) > sequence_length:
    for i in range(len(notes) - sequence_length):
        inputs.append([note_to_int[ch] for ch in notes[i:i + sequence_length]])
        outputs.append(note_to_int[notes[i + sequence_length]])

    X = np.reshape(inputs, (len(inputs), sequence_length, 1)) / float(len(pitches))
    y = keras.utils.to_categorical(outputs)
    print(f"✅ Sequence data prepared. Sequences: {X.shape[0]} (Significantly reduced from full dataset).")

    # Define and Train/Save Model (Reduced LSTM size)
    model = keras.Sequential([
        layers.LSTM(128, input_shape=(X.shape[1], X.shape[2]), return_sequences=True), # Reduced from 256
        layers.Dropout(0.3), layers.LSTM(128), # Reduced from 256
        layers.Dense(128, activation="relu"),
        layers.Dropout(0.3), layers.Dense(len(pitches), activation="softmax")
    ])
    model.compile(loss="categorical_crossentropy", optimizer="adam")

    print("🚀 Training model (5 epochs)... Expect this to be much faster!")
    # Training time is now much shorter due to reduced X and y size
    model.fit(X, y, epochs=5, batch_size=64)
    model.save(MODEL_PATH)
    print(f"✅ Model saved successfully to {MODEL_PATH}")

else:
    print("❌ ERROR: Not enough notes for sequencing.")
    sys.exit(0)
```

```python
%%writefile text2music_app.py
import streamlit as st
import music21 as m21
import pretty_midi
from scipy.io import wavfile
import numpy as np
import tempfile
import subprocess
import os

# ------------------------------------------------------------
# 🎹 STREAMLIT APP: User-Input Music Generator
# ------------------------------------------------------------
st.set_page_config(page_title="🎵 MUSIC GENERATOR", page_icon="🎶", layout="centered")

# --- Custom Centered Title ---
st.markdown(
    """
    <h1 style='text-align: center; color: white;'>🎵 MUSIC GENERATOR 🎵</h1>
    <p style='text-align: center;'>Create your own music using MIDI numbers!</p>
    """,
    unsafe_allow_html=True
)

st.info("💡 Enter MIDI numbers (C4=60, D4=62, E4=64, F4=65, G4=67, A4=69, B4=71). Example: `60,64,67,72`")

# --- Check FluidSynth Installation ---
soundfont_path = "/usr/share/sounds/sf2/FluidR3_GM.sf2"
if not os.path.exists(soundfont_path):
    st.warning("Installing FluidSynth soundfont... please wait ⏳")
    subprocess.run(["apt-get", "install", "-y", "fluidsynth", "fluid-soundfont-gm"], stdout=subprocess.PIPE)

# --- User Adjustable Controls ---
st.subheader("🎛️ Music Controls")
NOTE_DURATION = st.slider("🎚️ Note Duration (seconds per note):", 0.2, 2.0, 1.0, 0.1)
st.caption(f"Each note will play for {NOTE_DURATION:.1f} seconds.")

# --- Helper Functions ---
def create_midi(pitches, output_file="user_melody.mid"):
    """Convert MIDI pitch numbers into a MIDI file."""
    output_notes = []
    offset = 0
    for pitch in pitches:
        try:
            n = m21.note.Note(m21.pitch.Pitch(midi=pitch))
            n.quarterLength = NOTE_DURATION
            n.offset = offset
            output_notes.append(n)
            offset += NOTE_DURATION
        except:
            pass
    midi_stream = m21.stream.Stream(output_notes)
    midi_stream.write("midi", fp=output_file)
    return output_file

def midi_to_wav(midi_file):
    """Convert MIDI file to WAV using the system FluidSynth binary."""
    tmp_wav = tempfile.NamedTemporaryFile(delete=False, suffix=".wav")
    try:
        subprocess.run([
            "fluidsynth", "-ni", soundfont_path, midi_file,
            "-F", tmp_wav.name, "-r", "44100"
        ], check=True, stdout=subprocess.PIPE, stderr=subprocess.PIPE)
        return tmp_wav.name
    except Exception as e:
        st.error(f"Error converting MIDI to WAV: {e}")
        return None

# --- User Input Section ---
user_input = st.text_input(
    "🎹 Enter your own Music (comma-separated MIDI numbers):",
    "60,60,67,67,69,69,67"
)

if st.button("🎶 Generate Music"):
    try:
        user_midi = [int(n.strip()) for n in user_input.split(",") if n.strip().isdigit()]
        if not user_midi:
            st.error("❌ Please enter valid MIDI numbers (e.g., 60,64,67).")
        else:
            st.info("🎵 Creating your Music... please wait.")
            midi_path = create_midi(user_midi)
            wav_path = midi_to_wav(midi_path)
            if wav_path:
                st.success("✅ Music generated successfully!")
                st.audio(wav_path, format="audio/wav")
                st.download_button(
                    label="⬇️ Download MIDI File",
                    data=open(midi_path, "rb").read(),
                    file_name="user_melody.mid",
                    mime="audio/midi"
                )
            else:
                st.error("Failed to generate WAV. Please check logs.")
    except Exception as e:
        st.error(f"Error: {e}")

st.markdown("---")
st.caption("Developed by Dakshani,Ranjeetha and Krishna 🎶")
```
